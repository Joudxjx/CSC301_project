```python
%pip install pandas numpy

import pandas as pd
import heapq
import time
import numpy as np

def dijkstra(matrix, start):
    n = len(matrix)
    distances = [float('inf')] * n
    distances[start] = 0
    visited = [False] * n
    queue = [(0, start)]
    while queue:
        current_distance, current_node = heapq.heappop(queue)
        if visited[current_node]: continue
        visited[current_node] = True
        for neighbor in range(n):
            weight = matrix[current_node][neighbor]
            if weight > 0:
                new_distance = current_distance + weight
                if new_distance < distances[neighbor]:
                    distances[neighbor] = new_distance
                    heapq.heappush(queue, (new_distance, neighbor))
    return distances

df = pd.read_csv('distance_matrix_small (1).csv', index_col=0)
city_matrix = df.values.tolist()
cities = df.columns.tolist()

print("Setup Complete. Data loaded for:", cities)
```

```python
print("--- 4.2: Best Case Scenario ---")
start_time = time.perf_counter()

results_best = dijkstra(city_matrix, 0)

end_time = time.perf_counter()
print(f"Origin: {cities[0]} | Destination: {cities[0]}")
print(f"Calculated Distance: {results_best[0]}")
print(f"Execution Time: {end_time - start_time:.10f} seconds")
```

```python
print("--- 4.3: Average Case Scenario ---")
start_time = time.perf_counter()

avg_results = dijkstra(city_matrix, 0)

end_time = time.perf_counter()

print(f"Shortest distances from {cities[0]}:")
for i in range(len(cities)):
    print(f" -> {cities[i]}: {round(avg_results[i], 2)} km")

print(f"\nTotal Execution Time for 9 Cities: {end_time - start_time:.10f} seconds")
```

```python
print("--- 4.4: Worst Case Scenario (Stress Test) ---")

size = 200
worst_matrix = np.random.uniform(10, 500, size=(size, size)).tolist()
for i in range(size): worst_matrix[i][i] = 0 

start_time = time.perf_counter()

worst_results = dijkstra(worst_matrix, 0)

end_time = time.perf_counter()

print(f"Processed {size} cities and {size*size} possible paths.")
print(f"Worst Case Execution Time: {end_time - start_time:.10f} seconds")
```
